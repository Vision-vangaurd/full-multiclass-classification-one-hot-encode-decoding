```python
train_data = pd.read_csv('unzipped_content/data/train.csv')
train_data = train_data[['Filename','Male','Young','Middle_Aged','Senior']]
test_data = pd.read_csv('unzipped_content/data/test.csv')
test_data = test_data[['Filename']]
train_data = train_data.replace({1: 1, 0: 0.5, -1:0})
```

```python
from sklearn.model_selection import train_test_split
y_train, y_val = train_test_split(
    train_data,
    test_size=0.2,
    random_state=42,
)
```


<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
سپس نیاز است که تصاویر را از دایرکتوری های قرار گرفته در فایل های <code>test.csv</code> و <code>train.csv</code> بخوانید.
</font>
</p>

```python
import tensorflow as tf
import numpy as np

def df_to_tf_dataset(
    df,
    base_dir="/content/unzipped_content/data/image_data/",
    img_size=(256,256),
    batch_size=32,
    shuffle=True,
    augment=False,
    treat_neg_one_as_zero=True,
    cache=False
):
    paths = (base_dir + df.iloc[:, 0].astype(str)).values
    labels = df.iloc[:, 1:].values.astype(np.float32)

    if treat_neg_one_as_zero:
        labels[labels == -1] = 0.0

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))

    if shuffle:
        ds = ds.shuffle(buffer_size=len(paths), reshuffle_each_iteration=True)

    def _parse(path, label):
        image_bytes = tf.io.read_file(path)
        image = tf.image.decode_jpeg(image_bytes, channels=3)
        image = tf.image.convert_image_dtype(image, tf.float32)
        image = tf.image.resize(image, img_size)

        if augment:
            image = tf.image.random_flip_left_right(image)
            image = tf.image.random_brightness(image, max_delta=0.05)
            image = tf.image.random_contrast(image, lower=0.95, upper=1.05)

        return image, label

    ds = ds.map(lambda p, l: _parse(p, l), num_parallel_calls=tf.data.AUTOTUNE)

    if cache:
        ds = ds.cache()

    ds = ds.batch(batch_size).prefetch(tf.data.AUTOTUNE)
    return ds

train_ds = df_to_tf_dataset(y_train,
                            base_dir="/content/unzipped_content/data/image_data/",
                            img_size=(256,256),
                            batch_size=32,
                            shuffle=True,
                            augment=True,
                            treat_neg_one_as_zero=True,
                            cache=False)

val_ds = df_to_tf_dataset(y_val,
                          base_dir="/content/unzipped_content/data/image_data/",
                          img_size=(256,256),
                          batch_size=32,
                          shuffle=False,
                          augment=False,
                          treat_neg_one_as_zero=True,
                          cache=False)
test_ds = df_to_tf_dataset(test_data,
                          base_dir="/content/unzipped_content/data/image_data/",
                          img_size=(256,256),
                          batch_size=32,
                          shuffle=False,
                          augment=False,
                          treat_neg_one_as_zero=True,
                          cache=False)
```

```python
model = keras.Sequential()
inp = Input(shape = (256,256,3))
model.add(keras.layers.Conv2D(32,kernel_size=(3,3)))
model.add(keras.layers.BatchNormalization())
model.add(keras.layers.Activation('relu'))

model.add(keras.layers.MaxPool2D(pool_size=(2,2)))

model.add(keras.layers.Conv2D(64,kernel_size=(3,3)))
model.add(keras.layers.BatchNormalization())
model.add(keras.layers.Activation('relu'))

model.add(keras.layers.MaxPool2D(pool_size=(2,2)))

model.add(keras.layers.Conv2D(128,kernel_size=(3,3)))
model.add(keras.layers.BatchNormalization())
model.add(keras.layers.Activation('relu'))
model.add(keras.layers.MaxPool2D(pool_size=(2,2)))

model.add(keras.layers.Flatten())

model.add(keras.layers.Dense(units=256,activation='relu'))
model.add(keras.layers.Dropout(0.5))

model.add(keras.layers.Dense(units=128,activation='relu'))
model.add(keras.layers.Dense(units=4,activation='sigmoid'))
```
